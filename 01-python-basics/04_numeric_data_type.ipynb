### 🔢 2. Numeric Data Types (immutable)

* Numeric data types are used to store **numeric values**.
* In Python, the numeric types refer to the data types that represent numbers, which include 
   * int(0,100) 
   * float(10.0)
   * bool(true or false)
   * Complex Numbers(1a)

| **Data Type**   | **Category**   | **Example**     | **Definition** |
|-----------------|----------------|-----------------|----------------|
| **Numeric**     | int (Integer)  |   a = 10        | Holds whole numbers (e.g., 0, 100, -5). |
|                 | float (Float)  |   a = 10.5      | Numbers with decimal or fractional part. |
|                 | bool (Boolean) |   x = True      | Represents truth values: `True` or `False`. |
|                 | complex        |   a = 2 + 3j    | Numbers with real and imaginary parts. |



<br><br>

* **In Python, the numeric types refer to the data types that represent numbers, which include**
    - int(0,100)
    - float(10.0)

    - Complex Numbers(1a)

<br><br>

#### Creating Numeric Data Types

```python
numaric = 10
print("output of numaric value:", numaric )
print("output of numaric type:", type(numaric))
```

<br><br>


#### Creating float Types

```python
my_float = 10.5
print("output of float value:", my_float)
print("output of float type:", type(my_float))
```

<br><br>

#### Converting integer(numaric) to float

```python
numaric = 10
print("output of numaric value:", numaric )
print("output of numaric type:", type(numaric))

my_float = float(numaric)
print("output of float value:", my_float)
print("output of float type:", type(my_float))
```

<br><br>

#### Enter Runtime
- when you enter a value at runtime using input() function the default datatype is string

```python
numaric = int(input("Enter a number: "))       # input from user entered 5
print("output of numaric value:", numaric )
print("output of numaric type:", type(numaric))
```

<br><br>

#### Add two numaric values at runtime

```python
a = int(input("Enter a number: "))       # input from user entered 5
b = int(input("Enter a number: "))       # input from user entered 10

c = a + b

print("output of c value:", c )
print("output of c type:", type(c))
```

<br><br>

#### Adding two float values

```python
a = float(input("Enter num1:"))  # 10.5
b = float(input("Enter num2:"))  # 20.5

c = a + b

print("output of sum is:", c) 
```
