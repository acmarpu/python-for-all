### 🔑 Dictionary Data Type
--------------------------------------------------------------------------------------------------
#### What is dictionary : 
* A **dictionary** is an **unordered collection of key-value pairs**.  
* It stores elements in **pairs**, where each key is associated with a value. - Each key-value pair is called an **item**: `{key: value}` 

📌 **Key Points:**

   * **Ordered Collection:** As of Python 3.7+, dictionaries maintain the order of insertion, meaning the order in which you add key-value pairs is preserved.
   * **Keys are Immutable:** The keys must be of immutable types (e.g., strings, numbers, tuples). You cannot use mutable types (like lists) as keys.
   * **Unique Keys:** Dictionary keys must be unique. If you try to insert a duplicate key, it will update the existing key with the new value.
   * **Mutable Values:** Values can be of any data type, and they do not need to be unique.
   * **Creation:** You can create a dictionary using curly braces {} or by using the dict() constructor.
   * *Creating a Dictionary:*
   * **Empty dictionary:** Use {} or dict().
   * **Non-empty dictionary:** Use curly braces {} with key-value pairs separated by a colon (:).

<br><br>

#### Creating dictonary

```python
dict = {"code":"python", "server":"vm", "cloud": "azure" }
print("Output of Dictionary:", dict)
```

<br><br>

#### Accessing value from dictionary

```python
dict = {"code":"python", "server":"vm", "cloud": "azure" }
print("Output of single key value:" , dict["code"])
# print("Output of single key value:" , dict["age"])
```

<br><br>

#### Change or add values

```python
dict = {"code":"python", "server":"vm", "cloud": "azure" }
print("Output of Dictionary:", dict)

# Change
dict["cloud"] = "aws"
print("Output of Dictionary after change the key value:", dict)

# add
dict["data"] = "raw_data"
print("Output of Dictionary after adding the key value:", dict)
```

<br><br>

#### Delete or remove values

```python
dict = {"code":"python", "server":"vm", "cloud": "azure" }
del dict["server"]
print("Output of Dictionary after delete the key value:", dict)
```

<br><br>

#### Copy (Shallow Copy)

```python
dict = {"code":"python", "server":"vm", "cloud": "azure" }
print("Output of Dictionary:", dict)

dict1 = dict

print("Output of assigan value to dict1:", dict)


dict["server"] = "system"

print("Output of Dictionary:", dict)
print("Output of dict1:", dict1)
```

<br><br>

#### Items and keys and values
* The `keys()` method returns **all the keys** in a dictionary.

```python
dict = {"code":"python", "server":"vm", "cloud": "azure" }

print(dict.items())

# The `keys()` method returns **all the keys** in a dictionary.
print("output of keys:", dict.keys())

print("output of values:", dict.values())
```

<br><br>

#### Membership test
* Membership test applies only to keys.

```python
dict = {"code":"python", "server":"vm", "cloud": "azure" }

print(123 in dict)

print("server" in dict)
```

<br><br>

#### Len

```python
dict = {"code":"python", "server":"vm", "cloud": "azure" }
print(len(dict))
```

<br><br>

#### Pop

```python
dict = {"code":"python", "server":"vm", "cloud": "azure" }
value = dict.pop("server")
print(value)
```

<br><br>

#### Popitem

```python
dict = {"code":"python", "server":"vm", "cloud": "azure" }
dict.popitem()
print(dict)
```
