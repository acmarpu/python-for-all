### Modules
----------------------------------------------------------------------------------------------
##### What is a Module
* A module is just a Python file (.py)
* A **module** is like a single tool that contains reusable code.
* A **module** is a file that contains a set of code or *functions* that can be used in an application. A module can contain a single variable, a function, or a large amount of reusable code.
* Contains code (functions, classes, or variables) which you can import into another file.


##### Creating a Module

* To create a module, we write our code in a Python script and save it as a `.py` file. Create a file named `mymodule.py` inside your project folder. We can then write our functions, variables, or other Python code in this file.

##### Importing a Module
* To import the file we use the import keyword and the name of the file only.

##### import os Module
* The Python **`os` module** allows us to perform many **operating system-related tasks** automatically.

The `os` module provides functions for:
* Creating, changing, and removing directories (folders).
* Changing and identifying the **current working directory**.
* Fetching the contents of a directory.
* Working with files and directories.

```python
# import the module
import os

# Creating a directory
os.mkdir('directory_name')

# Changing the current directory
os.chdir('path')

# Getting current working directory
os.getcwd()

# Removing directory
os.rmdir()
```

##### import sys Module

* The **`sys` module** provides functions and variables that allow us to interact with the **Python runtime environment**.
* It is commonly used to work with **command-line arguments**, the Python interpreter, and other runtime information.
* The `sys.argv` function returns a **list of command-line arguments** passed to a Python script.

```python
import sys

# to exit sys
sys.exit()

# To know the largest integer variable it takes
sys.maxsize

# To know environment path
sys.path

# To know the version of python you are using
sys.version
```
