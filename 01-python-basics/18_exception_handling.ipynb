### Exception Handling
----------------------------------------------------------------------------------------------

* **What is Exception Handling?**
    - **`try-except` blocks**
    - **Handling different types of exceptions**
    - **`else` and `finally` blocks**
    - **Exception handling best practices**
- finally: finally block will be executed if we get error as well as if we don't get the error


##### 🐞 Types of Errors

- **Syntax errors** [Compile time Error] – Errors caused by incorrect Python syntax. 
- **Runtime errors (Exceptions)** – Errors that occur while the program is running.

```python
# Syntax Error
if 10 > 5
    print("Hello")
```

<br>

```python
### ⚠️ Runtime Errors
a = 10
b = 0

print(a / b)
```

```python
a = 10
b = 0

try:
 print(a / b)
except Exception as e:
 print("You cannot divide a number by zero:", e)
```

```python
a = 10
b = 2

try:
 print(a / b)
except Exception as e:
 print("You cannot divide a number by zero:", e)
```

```python
a = 10
b = 0

try:
 print("Hello, Good Morning!")
 print(a / b)
 print("Hello, Good Evening!")

except Exception as e:
 print("You cannot divide a number by zero:", e)
```

```python
a = 10
b = 5

try:
 print("Hello, Good Morning!")
 print(a / b)
 
except Exception as e:
 print("You cannot divide a number by zero:", e)
 print("Hello, Good Evening!")
```

```python
a = 10
b = 0

try:
 print("Hello, Good Morning!")
 print(a / b)
 
except Exception as e:
 print("You cannot divide a number by zero:", e)

finally:
 print("Hello, Good Evening!")
```

```python
a = 10
b = 0

try:
 print("Hello, Good Morning!")
 print(a / b)
 
except Exception as e:
 print("You cannot divide a number by zero:", e)

finally:
 print("Hello, Good Evening!")
```

#### NameError
print(value)
# Error: 'value' is not defined

#### IndexError
nums = [1, 2, 3]
print(nums[5])
# Error: List index out of range

#### KeyError
data = {"name": "Ashoka"}
print(data["age"])
# Error: Key 'age' not found in dictionary


#### ValueError
num = int("abc")
# Error: Invalid literal for int()


#### AttributeError
x = 10
x.append(5)
# Error: 'int' object has no attribute 'append'

#### IOError
f = open("nonexistent.txt", "r")
# Error: File not found

#### ZeroDivisionError
print(10 / 0)
# Error: Division by zero

#### ImportError
import non_existing_module
# Error: No module named 'non_existing_module'
