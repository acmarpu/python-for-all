### 📦 Tuple Data Type [immutable]
--------------------------------------------------------------------------------------------------
#### 1. what is tuple: 

* A **tuple** is a built-in data structure in Python, very similar to a **list**. However, unlike lists, **tuples are immutable** — their elements cannot be changed after creation.
   
📌 **Key Points:**
  * **Immutable:** Once a tuple is created, its elements cannot be modified, added, or removed.
  * **Ordered:** Like lists, tuples maintain the order of elements.
  * **Different Data Types:** A tuple can store elements of various data types (e.g., integers, strings, floats, etc.).
  * **Duplicates Allowed:** Tuples can have multiple occurrences of the same element.
  * **Parentheses:** Tuples are usually created using () (parentheses), though they are optional in some cases (e.g., creating a single element tuple requires a trailing comma).


* If you want to create a **tuple with only one element**, you must add a **trailing comma**.  
* Without the comma, Python will treat it as a normal value inside parentheses, not a tuple.
   

<br><br>

#### Creating tuple : ()

```python
tup = (10, 20, 30, 40)
print("Output of tup:", tup)
```

<br><br>

#### Tuple with mixed data types (heterogeneous)

```python
tup = (10, 3.5, "python", True)
print("Output of tup:", tup)
```

<br><br>

#### Creating tuple with one element:
* Must include a comma ( , ) after the element.

```python
tup = (10,)
print("Output of tup:", tup)
```

<br><br>

#### Accessing elements from tuple

```python
tup = (10, 20, 30, 40)
print(tup[1])
print(tup[-1])
```

<br><br>

#### Tuple slicing

```python
tup = (10, 20, 30, 40, 50, 60)
print(tup[1:4])
```

<br><br>

#### Tuple immutable
* Tuples cannot be modified (no item assignment).

```python
tup = ("mac",10,3.5,10,10)
print("Output of tup:", tup)
print("Output of type tup:", type(tup))
```

<br><br>

#### Tuple concatenation and multiplication
* adding one or more tuples in to single concatenation

```python
tup1 = (10, 20)
tup2 = (30, 40)

print("Output of total value:", tup1 + tup2)
```

<br><br>

#### Deleting a tuple

```python
tup_delete = (10, 20, 30)
del tup_delete
```

<br><br>

#### Count and index

```python
tup = (10, 20, 10, 30, 10)
print(tup.count(10))

print(tup.index(30))

tup = ("mac",10,3.5,10,10,"mac","mac",10,10)
print(tup.count("mac"))
```

<br><br>

#### Tuple membership tes

```python
tup = (10, 20, 30)
print(20 in tup)
print(40 not in tup)


tup = ("mac",10,3.5,10,10,"mac","mac",10,10)
print(100 in tup)
```

<br><br>

#### Len(), Max(), Min(), Sum()

```python
tup = ("mac",10,3.5,10,10,"mac","mac",10,10)
print(len(tup))

tup = (10,3.-5,10,10,10,10)
print(max((tup)))
print(min((tup))) 
print(sum((tup)))
```

<br><br>

#### Converting a string into tuple

```python
str = "python"
t = tuple(str)
print(t)
```

<br><br>

#### Converting a list into tuple

```python
list = [10, 20, 30]
tup = tuple(list)
print(tup)
```

<br><br>

#### Converting a tuple into string

```python
tup = ('p', 'y', 't', 'h', 'o', 'n')
str = ''.join(tup)
print(str)
```

<br><br>

#### Packing and unpacking

```python
tup = (10, 20, 30)
a, b, c = tup
print(a, b, c)
```

<br><br>

#### A nested tuple means a tuple inside another tuple.

```python
nested_tuple = (1, 2, (3, 4), 5)

print(nested_tuple[0])

print(nested_tuple[1])

print(nested_tuple[2])

print(nested_tuple[3])
```
