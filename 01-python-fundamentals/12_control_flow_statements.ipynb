### 🔀 Control Flow Statements
----------------------------------------------------------------------------------------------
* **Control flow statements** determine the order in which statements are executed in a program.
* They control the flow of program execution based on conditions, loops, or other control statements.
* Conditional statements evaluate a condition:
  * If the condition is **True**, the corresponding block of code is executed.
  * If the condition is **False**, the block is skipped or an alternative block is executed.

✅ All loops are control flow statements,
but ❌ not all control flow statements are loops.

#### 📌 Types of Control Flow Statements

* Conditional Statements
* Loop Statements (Iterative)
* Transfer Statements

----------------------------------------------------------------------------------------------
#### 1) Conditional statements: (Decision Making)
----------------------------------------------------------------------------------------------
* **if**        = Executes a code block only if a specified condition evaluates to true.
* **else**      = Runs a fallback block if all previous conditions evaluate to false.
* **nested if** = if statements inside if statements. This is called nested if statements.
* **elif**      = Short for "else if"; checks another condition if the previous conditions were false.

----------------------------------------------------------------------------------------------
#### 2) Iterative Statements (Loops)
----------------------------------------------------------------------------------------------
* **for**       = Iterates over a sequence (such as a list, tuple, dictionary, set, or string) or a specified range.
* **whle**      = Continues execution as long as a specified condition remains true

----------------------------------------------------------------------------------------------
#### 3) Transfer (Jump) Statements
----------------------------------------------------------------------------------------------
* **break**    = Terminates the loop prematurely and jumps out of it completely.
* **continue** = Skips the remainder of the current loop iteration and moves directly to the next one.
* **pass**     = Acts as a null statement or placeholder where code is syntactically required but no action is needed
