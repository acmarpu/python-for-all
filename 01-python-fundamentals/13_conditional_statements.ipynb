### 1.❓Conditional Statements
----------------------------------------------------------------------------------------------
* These statements allow the program to evaluate a condition and choose different execution paths based on whether that condition is true or false


* **if** Based on the condition. it decides whether the block of code will allows for execution or not.
* The `if` keyword is used to test specific conditions.  
  - If the condition evaluates to **True**, the control enters the `if` block.  
  - If the condition evaluates to **False**, the control skips the `if` block and continues with the rest of the code.
* **else**      = Runs a fallback block if all previous conditions evaluate to false.
* **nested if** = if statements inside if statements. This is called nested if statements.
* **elif**      = Short for "else if"; checks another condition if the previous conditions were false.

* whenever performing checks and validations 
* All conditional statemnt boolen output
* Whenever working with conditionals statement we will use Logical statement and Comparison

### Conditional Statement with Dictionary

```python
user_roles = {
    "alice": "admin",
    "bob": "editor",
    "chris": "viewer"
}

user_name = "bob"

if user_name in user_roles:
    role = user_roles[user_name] 
    
    if role == "admin":
        print(f"{user_name} has full access.")

    elif role == "editor":
        print(f"{user_name} can edit content.")

    elif role == "viewer":
        print(f"{user_name} can view content.")

    else:
     print(f"{user_name} is not found in the user roles.")
```

### 🪜 Nested if
* In simple terms: if inside another if is called a nested if statement.

```python
user_name = "admin"
password = "admin123"

if user_name == "admin":
    if password == "admin123":
        print("Login successful.")
    else:
        print("Incorrect password.")
else:
    print("Invalid username.")
```

### 🔂 elif

* If one of the conditions is true, the remaining conditions will not be tested.

* The elif statement allows you to check multiple conditions. As soon as one condition is true, its block of code executes, and the rest are skipped.

* Using nested if statements (first example) makes the code longer and harder to read.

```python
i = int(input("enter the number"))  # Number entered: 3
if i == 1:
    print("one")
else:
    if i == 2:
        print(f" Output: {i} is two")
    else:
        if i == 3:
            print(f" Output: {i} is three")
        else:
            if i == 4:
                print(f" Output: {i} is four")
            else:
                print(f" Output: {i} is invalid")
```

* The elif ladder checks conditions one by one — once a condition is True, the rest are skipped.


```python
i = int(input("Enter a number: "))   # Define i before using it

if i == 1:
    print("one")
elif i == 2:
    print("two")
elif i == 3:
    print("three")
elif i == 4:
    print("four")
else:
    print("invalid number")
```
